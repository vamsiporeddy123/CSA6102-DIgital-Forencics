```python
import hashlib

def compute_hashes(data: bytes):
    return {
        "MD5": hashlib.md5(data).hexdigest(),
        "SHA1": hashlib.sha1(data).hexdigest(),
        "SHA256": hashlib.sha256(data).hexdigest(),
    }

original_data = b"hello"
tampered_data = b"Hello"


hashes_original = compute_hashes(original_data)
hashes_tampered = compute_hashes(tampered_data)
hashes_original_repeat = compute_hashes(original_data)

print("Original Data (hello):")
print(hashes_original)

print("\nTampered Data (Hello):")
print(hashes_tampered)

print("\nRepeated Hash of Original Data:")
print(hashes_original_repeat)

print("\nConsistency Check:")
print("Original hashes match repeated hashes:", hashes_original == hashes_original_repeat)

print("Original and tampered hashes are the same:", hashes_original == hashes_tampered)
```

```python
def recover_deleted_messages(sms_table):
    recoverable = [
        row for row in sms_table
        if row["is_deleted"] and not row["overwritten"]
    ]
    return recoverable


def summarize_table(sms_table):
    active = [r for r in sms_table if not r["is_deleted"]]
    recoverable = recover_deleted_messages(sms_table)
    lost = [
        r for r in sms_table
        if r["is_deleted"] and r["overwritten"]
    ]

    return {
        "active": len(active),
        "recoverable": len(recoverable),
        "permanently_lost": len(lost)
    }


sms_table = [
    {
        "id": 1,
        "sender": "Alice",
        "message": "Meeting at 10 AM",
        "is_deleted": False,
        "overwritten": False
    },
    {
        "id": 2,
        "sender": "Bob",
        "message": "Your OTP is 123456",
        "is_deleted": True,
        "overwritten": False
    },
    {
        "id": 3,
        "sender": "Charlie",
        "message": "Happy Birthday!",
        "is_deleted": True,
        "overwritten": True
    },
    {
        "id": 4,
        "sender": "David",
        "message": "Call me back",
        "is_deleted": False,
        "overwritten": False
    }
]


recoverable = recover_deleted_messages(sms_table)
summary = summarize_table(sms_table)

print("=" * 50)
print("SMS FORENSIC RECOVERY REPORT")
print("=" * 50)

for msg in recoverable:
    print("Message ID :", msg["id"])
    print("Sender     :", msg["sender"])
    print("Message    :", msg["message"])
    print()

print("Summary")
print("Active Messages      :", summary["active"])
print("Recoverable Messages :", summary["recoverable"])
print("Permanently Lost     :", summary["permanently_lost"])

# Test Cases
assert len(recoverable) == 1
assert recoverable[0]["sender"] == "Bob"
assert summary["active"] == 2
assert summary["recoverable"] == 1
assert summary["permanently_lost"] == 1

print("\nAll test cases passed.")
```

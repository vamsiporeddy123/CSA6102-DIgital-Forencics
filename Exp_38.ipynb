```python
import hashlib
import math
from collections import Counter

def sha256_hash(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()

def byte_entropy(data: bytes) -> float:
    if not data:
        return 0.0

    counts = Counter(data)
    length = len(data)

    return -sum(
        (count / length) * math.log2(count / length)
        for count in counts.values()
    )

def static_analyze(data: bytes, entropy_threshold=7.5):

    entropy = byte_entropy(data)

    return {
        "sha256": sha256_hash(data),
        "size_bytes": len(data),
        "entropy": round(entropy, 2),
        "likely_packed": entropy >= entropy_threshold,
    }


sample_data = b"""
This is a sample executable file used for static malware analysis.
It contains normal readable text for demonstration purposes.
"""


report = static_analyze(sample_data)

print("=" * 60)
print("STATIC MALWARE ANALYSIS REPORT")
print("=" * 60)

print(f"SHA-256 Hash      : {report['sha256']}")
print(f"File Size         : {report['size_bytes']} bytes")
print(f"Entropy           : {report['entropy']:.2f}")
print(f"Likely Packed     : {report['likely_packed']}")

if report["likely_packed"]:
    print("\nAssessment : High entropy detected. File may be packed or encrypted.")
else:
    print("\nAssessment : Entropy is within the normal range.")


assert report["size_bytes"] == len(sample_data)
assert report["sha256"] == sha256_hash(sample_data)
assert report["entropy"] >= 0

print("\n" + "=" * 60)
print("TEST CASE RESULTS")
print("=" * 60)
print("✓ Test Case 1 Passed")
print("✓ Test Case 2 Passed")
print("✓ Test Case 3 Passed")

print("\nAll test cases passed successfully.")
```

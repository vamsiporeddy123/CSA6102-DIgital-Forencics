```python
import math
from collections import Counter

def shannon_entropy(s):
    if not s:
        return 0.0

    counts = Counter(s)
    length = len(s)

    return -sum(
        (count / length) * math.log2(count / length)
        for count in counts.values()
    )

def detect_dns_tunneling(queries, length_threshold=20, entropy_threshold=3.5):
    """
    Detect DNS tunneling by checking:
    1. Long subdomain label
    2. High Shannon entropy
    """

    flagged = []

    for query in queries:

        label = query.split(".")[0]
        entropy = shannon_entropy(label)

        if len(label) >= length_threshold and entropy >= entropy_threshold:

            flagged.append({
                "query": query,
                "label_length": len(label),
                "entropy": round(entropy, 2)
            })

    return flagged


queries = [
    "www.google.com",
    "mail.microsoft.com",
    "abcdefghijklmnopqrstuvwxyz123456.example.com",
    "a8x9k2m4q7z1p5n8r6v3t0y1u2.example.org",
    "facebook.com"
]


results = detect_dns_tunneling(queries)

print("=" * 60)
print("DNS TUNNELING DETECTION REPORT")
print("=" * 60)

if results:
    for item in results:
        print(f"\nQuery          : {item['query']}")
        print(f"Label Length   : {item['label_length']}")
        print(f"Entropy Score  : {item['entropy']}")
        print("Status         : SUSPICIOUS")
else:
    print("No DNS tunneling detected.")


assert abs(shannon_entropy("aaaa")) == 0.0
assert shannon_entropy("abcd1234") > 2.5

flagged = detect_dns_tunneling(queries)

assert len(flagged) >= 2

print("\n" + "=" * 60)
print("TEST CASE RESULTS")
print("=" * 60)
print("✓ Test Case 1 Passed")
print("✓ Test Case 2 Passed")
print("✓ Test Case 3 Passed")
print("\nAll test cases passed successfully.")
```
